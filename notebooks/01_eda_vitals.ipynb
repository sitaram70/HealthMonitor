# 01 — EDA on Simulated Vitals

This notebook explores the simulated heart rate and temperature time‑series and the `risk_event` labels.


```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/vitals/vitals_train.csv', parse_dates=['timestamp'])
df.set_index('timestamp')[['heart_rate_bpm','temp_c']].plot(figsize=(10,3))
plt.title('Heart Rate & Temperature'); plt.show()
df['risk_event'].rolling(60).mean().plot(figsize=(10,2))
plt.title('Risk Event Rolling Mean'); plt.show()
df.head()
```
